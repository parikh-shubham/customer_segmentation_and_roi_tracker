```python
import pandas as pd
import numpy as np
from pathlib import Path
import random
from datetime import datetime, timedelta
```

```python
np.random.seed(42)
random.seed(42)
```

```python
# Customers Table

N_CUSTOMERS = 5000
N_ORDERS = 20000
N_CAMPAIGNS = 8
START_DATE = datetime(2022, 1, 1)
END_DATE = datetime(2024, 12, 31)

date_range = (END_DATE - START_DATE).days

CHANNELS = ['Email', 'Paid Social', 'SMS', 'Google Search', 'Organic']
CITITES = ['Mumbai', 'Delhi', 'Bengaluru', 'Pune', 'Hyderabad', 'Chennai']
CATEGORIES = ['Electronics', 'Apparel', 'Home & Kitchen', 'Beauty', 'Sports']
```

```python
customers = pd.DataFrame({
    'customer_id': range(1001, 1001 + N_CUSTOMERS),
    'name': [f'customer_{i}' for i in range(N_CUSTOMERS)],
    'email': [f'customer{i}.gmail.com' for i in range(N_CUSTOMERS)],
    'city': np.random.choice(CITITES, N_CUSTOMERS),
    'signup_channel': np.random.choice(CHANNELS, N_CUSTOMERS),
    'signup_date': [(START_DATE + timedelta(days=np.random.randint(0, date_range))).strftime('%Y-%m-%d') for _ in range(N_CUSTOMERS)],
    'age': np.random.randint(18, 65, N_CUSTOMERS),
    'gender': np.random.choice(['M', 'F', 'Others'], N_CUSTOMERS, p=[0.48, 0.48, 0.04])
})
```

```python
# Orders Table

order_customer_ids = np.random.choice(customers['customer_id'], N_ORDERS)
order_amounts = np.round(np.random.lognormal(mean=5.5, sigma=0.8, size=N_ORDERS), 2)
order_dates = [(START_DATE + timedelta(days=np.random.randint(0, date_range))).strftime('%Y-%m-%d') for _ in range(N_ORDERS)]

# Injecting some noise
order_amounts[:100] = -order_amounts[:100]
order_amounts[100:150] = np.nan

orders = pd.DataFrame({
    'order_id': range(5001, 5001 + N_ORDERS),
    'customer_id': order_customer_ids,
    'order_date': order_dates,
    'amount': order_amounts,
    'category': np.random.choice(CATEGORIES, N_ORDERS),
    'channel': np.random.choice(CHANNELS, N_ORDERS),
    'status': np.random.choice(['Completed', 'Completed', 'Completed', 'Returned', 'Cancelled'], N_ORDERS)
})

# Adding duplicate rows
orders = pd.concat([orders, orders.sample(200)], ignore_index=True)
```

```python
# Campaigns Table

campaigns = pd.DataFrame({
    'campaign_id': range(1, N_CAMPAIGNS + 1),
    'campaign_name': ['Summer Sale 2022', 'Festive Offer 2022', 'New Year 2023', 'Loyalty Reward Q2 2023', 'Back to school 2023', 'Festive Offer 2023', 'Re-engagement Jan 2024', 'Spring Sale 2024'],
    'channel': np.random.choice(CHANNELS[:-1], N_CAMPAIGNS),
    'start_date': ['2022-06-01', '2022-10-01', '2023-01-01', '2023-04-01', '2023-07-01', '2023-10-01', '2024-01-15', '2024-03-01'],
    'end_date': ['2022-06-30', '2022-11-15', '2023-01-31', '2023-04-30', '2023-07-31', '2023-11-30', '2024-02-15', '2024-03-31'],
    'budget_inr': np.random.choice([50000, 75000, 100000, 150000, 200000], N_CAMPAIGNS),
    'target_segment': np.random.choice(['All', 'New', 'Loyal', 'At-Risk', 'Churned'], N_CAMPAIGNS)
})
```

```python
# Campaigns Responses Table

n_resp = 15000

campaign_responses = pd.DataFrame({
    'reponses_id': range(1, n_resp + 1),
    'campaign_id': np.random.choice(campaigns['campaign_id'], n_resp),
    'customer_id': np.random.choice(customers['customer_id'], n_resp),
    'sent': 1,
    'opended': np.random.choice([0,1], n_resp, p=[0.35, 0.65]),
    'clicked': np.random.choice([0,1], n_resp, p=[0.70, 0.30]),
    'converted': np.random.choice([0,1], n_resp, p=[0.82, 0.18]),
    'revenue_generated': np.where(np.random.random(n_resp) > 0.82, np.round(np.random.lognormal(5.2, 0.7, n_resp), 2), 0)
})
```
